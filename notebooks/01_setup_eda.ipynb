# 1. Context

Sexual self-concept refers to the multidimensional cognitive, affective, and evaluative perceptions an individual holds about their own sexual feelings, behaviors, and identity. Such tendencies regarding an individual's sexuality can be measured using the Multidimensional Sexual Self-Concept Questionnaire (MSSCQ; Looti, 2025; Snell, 2001). This questionnaire consists of 100 questions where participants answer with a 5 point Likert-type scale (1: "not at all characteristic of me", 5: "very characteristic of me").

In this project, I use unsupervised machine learning to identify phenotypic attitude profiles regarding sexuality. To do so, I use a dataset retrieved from Kaggle (https://www.kaggle.com/datasets/yamqwe/multidimensional-sexual-selfconcept/data?select=MSSCQ) containing 17685 responses to the MSSCQ.

References:
* Looti, M. (2025). _The Multidimensional Sexual Self-Concept Questionnaire (MSSCQ)_. Psychological Scales & Instruments Database. https://db.arabpsychology.com/scales/the-multidimensional-sexual-self-concept-questionnaire-msscq/
* Snell Jr, W. E. (2001). Measuring multiple aspects of the sexual self-concept: The Multidimensional Sexual Self-Concept Questionnaire. _New Directions in the Psychology of Human Sexuality: Research and Theory, 17_.

To compute subscale scores, items are averaged over each subscale (Snell, 2001).

Following recommendations by Wu, Gu, and Fukui (2022), subscale scores were computed as the mean of available items when at least three of the five items were completed (i.e., when at least 50% responses were complete). Subscale scores with fewer than three completed items were treated as missing and subsequently imputed using the median.

```python
# Compute subscale scores for each subscale as described above

# Create disctionary
subscales = {
"Sexual_Anxiety": ["Q1", "Q21", "Q41", "Q61", "Q81"],
"Sexual_Self_Efficacy": ["Q2", "Q22", "Q42", "Q62", "Q82"],
"Sexual_Consciousness": ["Q3", "Q23", "Q43", "Q63", "Q83"],
"Motivation_to_Avoid_Risky_Sex": ["Q4", "Q24", "Q44", "Q64", "Q84"],
"Chance_Luck_Sexual_Control": ["Q5", "Q25", "Q45", "Q65", "Q85"],
"Sexual_Preoccupation": ["Q6", "Q26", "Q46", "Q66", "Q86"],
"Sexual_Assertiveness": ["Q7", "Q27", "Q47", "Q67", "Q87"],
"Sexual_Optimism": ["Q8", "Q28", "Q48", "Q68", "Q88"],
"Sexual_Problem_Self_Blame": ["Q9", "Q29", "Q49", "Q69", "Q89"],
"Sexual_Monitoring": ["Q10", "Q30", "Q50", "Q70", "Q90"],
"Sexual_Motivation": ["Q11", "Q31", "Q51", "Q71", "Q91"],
"Sexual_Problem_Management": ["Q12", "Q32", "Q52", "Q72", "Q92"],
"Sexual_Esteem": ["Q13", "Q33", "Q53", "Q73", "Q93"],
"Sexual_Satisfaction": ["Q14", "Q34", "Q54", "Q74", "Q94"],
"Power_Other_Sexual_Control": ["Q15", "Q35", "Q55", "Q75", "Q95"],
"Sexual_Self_Schemata": ["Q16", "Q36", "Q56", "Q76", "Q96"],
"Fear_of_Sex": ["Q17", "Q37", "Q57", "Q77", "Q97"],
"Sexual_Problem_Prevention": ["Q18", "Q38", "Q58", "Q78", "Q98"],
"Sexual_Depression": ["Q19", "Q39", "Q59", "Q79", "Q99"],
"Internal_Sexual_Control": ["Q20", "Q40", "Q60", "Q80", "Q100"]
}

# Compute subscale scores while requiring at least 3 out of 5 items to be answered fro each subscale
subscale_scores = pd.DataFrame(index=X_train.index)

for name, items in subscales.items():
    score = X_train[items].mean(axis=1)
    enough_answers = X_train[items].count(axis=1) >= 3
    subscale_scores[name] = score.where(enough_answers)
```

```python
# Check overall structure for 5 first observations
print(f'=== 5 first observations === \n {subscale_scores.head(5)} \n')

# Check NA
print(f'=== NA count for subscales === \n {subscale_scores.isna().sum()}')
```

```python
# Correlation matrix on X_train
pearson_matrix_subscales = subscale_scores.corr(numeric_only=True)

# Mask to only show inferior triangle
mask = np.triu(np.ones_like(pearson_matrix_subscales))

# Visualization of correlation matrix
fig, ax = plt.subplots(figsize=(13, 10))
sns.heatmap(
    pearson_matrix_subscales,
    vmin=-1,
    vmax=1,
    mask=mask,
    annot=True,
    fmt=".2f",
    linewidths=0.5,
    ax=ax,
    cmap="coolwarm"
)
plt.title("Pearson Correlation Matrix - Subscale scores")
plt.tight_layout()
plt.show()
```

The subscale correlation matrix reveals a relatively homogeneous pattern of predominantly weak-to-moderate correlations (|r| ≈ 0.10 – 0.40), with a small number of stronger positive and negative correlations (|r| ≈ 0.60 – 0.80). The strongest positive correlations were observed among Sexual Self-Efficacy, Sexual Esteem, and Sexual Satisfaction, while Sexual Anxiety and Sexual Depression showed substantial negative correlations with these constructs. Overall, the pattern suggests that the subscales assess related but distinct dimensions of sexual attitudes and experiences, supporting the use of dimensionality reduction techniques such as PCA while indicating that the subscales are not redundant. We can therefore expect respondents to be distinguishable by different combinations of subscale scores, allowing a clustering model to identify interpretable attitude profiles rather than clusters driven by a single dominant variable.

## 10.2. Pre-processing pipeline: impute, one-hot encode, scale, pca

Here, I will build two different preprocessing pipelines combining different scalers:
- StandardScaler
- RobustScaler

### 10.2.1. Create general pre-processing pipeline

```python
# Define column groups
subscale_cols = list(subscales.keys())
numeric_cols = subscale_cols + ["age"]
categorical_cols = ["gender"]

# Create pre-processing pipeline
def build_preprocessor(scaler):

    return ColumnTransformer(
        transformers=[
            (
            "numeric",
            Pipeline([ # Sub-pipeline for numeric variables (questions, age)
                ("imputer", SimpleImputer(strategy="median")), # Impute with median
                ("scaler", scaler) # Standardization
            ]),
            numeric_cols
            ),
            (
            "gender",
            Pipeline([
                ("imputer", SimpleImputer(strategy="most_frequent")), # Impute with mode
                ("onehot", OneHotEncoder(drop="first")) # One-hot encoding
            ]),
            categorical_cols
            )
        ]
    )
```

### 10.2.2. Apply pre-processing pipeline

#### 10.2.2.1. StandardScaler

```python
# Use previously defined general pre-processing pipeline with StandardScaler
standard_preprocessor = build_preprocessor(
    StandardScaler()
)
```

#### 10.2.2.2. RobustScaler

```python
# Use previously defined general pre-processing pipeline with RobustScaler
robust_preprocessor = build_preprocessor(
    RobustScaler()
)
```
